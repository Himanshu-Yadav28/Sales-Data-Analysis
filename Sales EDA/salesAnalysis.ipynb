```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("sales_data_sample.csv",encoding='latin1')
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.head(5)
```

```python
df.duplicated().sum()
```

```python
df['ORDERDATE']=pd.to_datetime(df['ORDERDATE'])
```

```python
df['ORDERDATE']
```

```python
df['Revenue']=df['QUANTITYORDERED']*df['PRICEEACH']
```

```python
df['Revenue']
```

```python
monthly_sales = df.groupby(df['ORDERDATE'].dt.to_period("M"))['Revenue'].sum()
```

```python
monthly_sales
```

```python
top_product = df.groupby(df['PRODUCTLINE'])['Revenue'].sum()
```

```python
top_product
```

```python
total_revenue = df['Revenue'].sum()
```

```python
total_revenue
```

```python
def plot_sales_trends():
    plt.figure(figsize=(12, 6))
    sns.lineplot(x=monthly_sales.index.astype(str), y=monthly_sales.values, marker="o", color="royalblue", linewidth=2.5)
    plt.title("Monthly Sales Trend", fontsize=14, fontweight='bold', color='darkblue')
    plt.xlabel("Month", fontsize=12)
    plt.ylabel("Revenue ($)", fontsize=12)
    plt.xticks(rotation=45, fontsize=10)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    sns.despine()
    #st.pyplot(plt)
```

```python
plot_sales_trends()
```

```python
def plot_top_products():
    plt.figure(figsize=(12, 6))
    sns.barplot(x=top_product.head(10).index, y=top_product.head(10).values, palette="viridis")
    plt.xticks(rotation=45, fontsize=10)
    plt.title("Top 10 Best-Selling Product Lines", fontsize=14, fontweight='bold', color='darkblue')
    plt.xlabel("Product Line", fontsize=12)
    plt.ylabel("Revenue ($)", fontsize=12)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    sns.despine()
    #st.pyplot(plt)
```

```python
plot_top_products()
```
